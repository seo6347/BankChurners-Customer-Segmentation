## 범주형 변수 인코딩
범주형 변수를 머신러닝 모델에 사용할 수 있도록 수치형으로 변환합니다.

```python
from sklearn.preprocessing import LabelEncoder

# 범주형 변수 확인
cat_cols = df.select_dtypes(include='object').columns.tolist()
print('범주형 변수:', cat_cols)

# Attrition_Flag 이진 인코딩 (이탈=1, 기존=0)
df['Attrition_Flag'] = df['Attrition_Flag'].map(
    {'Attrited Customer': 1, 'Existing Customer': 0})

# Gender 이진 인코딩
df['Gender'] = df['Gender'].map({'M': 1, 'F': 0})

# 나머지 범주형 변수 Label Encoding
le = LabelEncoder()
for col in ['Education_Level', 'Marital_Status', 'Income_Category', 'Card_Category']:
    df[col] = le.fit_transform(df[col])

print('인코딩 완료')
df.head()
```

## log 변환 파생변수 생성

우측 꼬리 분포(right-skewed)를 가진 수치형 변수에 `log1p` 변환을 적용합니다.

| 원본 변수 | 변환 변수 | 변환 이유 |
|---|---|---|
| `Credit_Limit` | `Credit_Limit_log` | 신용 한도는 소수 고액 고객으로 분포가 오른쪽으로 치우침 |
| `Total_Trans_Amt` | `Total_Trans_Amt_log` | 총 거래 금액도 극단값이 많아 정규화 필요 |
| `Avg_Open_To_Buy` | `Avg_Open_To_Buy_log` | 평균 가용 한도(신용한도 - 잔액)도 유사한 분포 |
| `Total_Revolving_Bal` | `Total_Revolving_Bal_log` | 총 회전 잔액(카드 미결제 잔액)은 0이 많아 log1p가 적합 |

> `log1p(x) = log(1+x)` : 0값도 안전하게 변환 가능

```python
import numpy as np

skewed_cols = ['Credit_Limit', 'Total_Trans_Amt', 'Avg_Open_To_Buy', 'Total_Revolving_Bal']

for col in skewed_cols:
    df[col + '_log'] = np.log1p(df[col])

print('log transform 완료:', [c + '_log' for c in skewed_cols])
```

## Age_Group 파생변수 설명

`Customer_Age`를 10세 단위 구간으로 나눠 `Age_Group` 파생변수를 생성했습니다.

| 구간 | 레이블 | 의미 |
|---|---|---|
| 20~30세 | `20대` | 사회 초년생, 신용카드 첫 사용층 |
| 30~40세 | `30대` | 소비 활동 활발, 카드 주 사용층 |
| 40~50세 | `40대` | 자산 형성 시기, 고액 한도 사용 가능성 |
| 50~60세 | `50대` | 소득 피크 구간, 보수적 소비 패턴 |
| 60세 이상 | `60대+` | 은퇴 준비 시기, 사용 빈도 감소 가능성 |

> 연속형 나이보다 구간 변수로 변환하면 **연령대별 이탈 패턴 시각화 및 그룹 비교**가 용이합니다.

```python
# Clustering에 사용할 변수 (행동 기반)
clustering_features = [
    'Total_Trans_Ct',
    'Total_Trans_Amt_log',
    'Months_Inactive_12_mon',
    'Total_Ct_Chng_Q4_Q1',
    'Total_Amt_Chng_Q4_Q1',
    'Total_Relationship_Count',
    'Months_on_book',
    'Contacts_Count_12_mon',
    'Credit_Limit_log',
    'Total_Revolving_Bal_log',
    'Avg_Utilization_Ratio'
]

# Profiling에 사용할 변수 (인구통계 + 이탈 여부)
profiling_features = [
    'Attrition_Flag',
    'Customer_Age',
    'Gender',
    'Education_Level',
    'Marital_Status',
    'Income_Category',
    'Card_Category',
    'Dependent_count'
]

df_cluster = df[clustering_features].copy()
df_profile = df[profiling_features].copy()

print('Clustering 변수 수:', len(clustering_features))
print('Profiling 변수 수:', len(profiling_features))
```

```python
from sklearn.preprocessing import StandardScaler

# 스케일러 정의 및 적용
scaler = StandardScaler()
# 원본 df에 스케일링된 컬럼들을 직접 추가하거나 덮어씁니다.
scaled_data = scaler.fit_transform(df[clustering_features])
scaled_cols = [c + '_scaled' for c in clustering_features]
df[scaled_cols] = scaled_data

print('스케일링 완료 및 df 반영 성공')
display(df[scaled_cols].describe().round(2))
```
